```python
import numpy as np
```

```python
# SVM classifier from scratch

class SVM_classifier():
  #initiating hyper parameters
  def __init__(self, learning_rate, no_of_iterations, lambda_parameter):
    self.learning_rate = learning_rate
    self.no_of_iterations = no_of_iterations
    self.lambda_parameter = lambda_parameter

  #fitting the dataset into SVMN classifier

  def fit(self, X, Y):

    self.m, self.n = X.shape

    self.w = np.zeros(self.n)
    self.b = 0
    self.X = X
    self.Y = Y

    for i in range(self.no_of_iterations):
      self.update_weights()

  #function for updating the weights and bias
  def update_weights(self):
    y_label = np.where(self.Y <= 0, -1, 1)
    for index, x_i in enumerate(self.X):
      condition = y_label[index] * (np.dot(x_i, self.w) - self.b) >= 1
      if (condition == True):
        dw = 2 * self.lambda_parameter * self.w
        db = 0
      else:
        dw = 2 * self.lambda_parameter * self.w - np.dot(x_i, y_label[index])
        db = y_label[index]


      #updating the weights

      self.w = self.w - self.learning_rate * dw
      self.b = self.b - self.learning_rate * db




  #predict the label for a given input value
  def predict(self, X):
    output = np.dot(X, self.w) - self.b
    predicted_labels = np.sign(output)
    y_hat = np.where(predicted_labels <= -1, 0, 1)
    return y_hat
```

```python
#implementing SVM classifier model

import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('diabetes.csv')
```

```python
#splitting datset into features and target

X = df.drop(columns = 'Outcome', axis = 1)
Y = df['Outcome']
```

```python
# data standardization

scaler = StandardScaler()
scaler.fit(X)
standardized_data = scaler.transform(X)
X= standardized_data
```

```python
#splitting into training and test dataset

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, stratify = Y, random_state = 2)
print(X.shape, X_train.shape, X_test.shape)
```

```python
model = SVM_classifier(learning_rate = 0.001, no_of_iterations = 1000, lambda_parameter = 0.01)
```

```python
model.fit(X_train, Y_train)
```

```python

#accuracy score on training set

X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(Y_train, X_train_prediction)
print('Accuracy score on training data : ', training_data_accuracy)

```

```python
#accuracy score on test set

X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(Y_test, X_test_prediction)
print('Accuracy score on test data : ', test_data_accuracy)
```
