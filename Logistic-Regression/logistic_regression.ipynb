```python
import numpy as np
```

```python
#logistic regression

class logisticRegression():
  def __init__(self, learningRate, iterations):
    self.learningRate = learningRate
    self.iterations = iterations


  def fit(self, X, Y):
    self.m, self.n = X.shape
    self.w = np.zeros(self.n)
    self.b = 0
    self.X = X
    self.Y = Y

    #implementing gradient descent
    for i in range(self.iterations):
      self.update_weights()

  def update_weights(self):

    #calculate gradients
    y_pred = self.predict(self.X)
    dw = (1/self.m)*np.dot(self.X.T, (y_pred - self.Y))
    db = (1/self.m)*np.sum(y_pred - self.Y)

    #updating weights

    self.w = self.w - self.learningRate*dw
    self.b = self.b - self.learningRate*db


  def predict(self, X):

    z = (X.dot(self.w) + self.b)
    y_pred = 1 / (1 + np.exp(-z))
    y_pred = np.where(y_pred > 0.5, 1, 0)
    return y_pred
```

```python
#implementing logistic regression model
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('diabetes.csv')
```

```python
df.isnull().sum()
```

```python
x = df.drop('Outcome', axis=1)
y = df['Outcome']
```

```python
#stanadardization
scaler = StandardScaler()
x = scaler.fit_transform(x)
```

```python
print(y)
```

```python
#splitting it into training amd test set
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=2)
```

```python
print(x.shape, x_train.shape, x_test.shape)
```

```python
model = logisticRegression(learningRate=0.01, iterations=1000)
```

```python
model.fit(x_train, y_train)
```

```python
#evaluation
y_pred = model.predict(x_train)
train_acc = accuracy_score(y_train, y_pred)
print(train_acc)
```

```python
#accuracy score on test set

y_pred = model.predict(x_test)
test_acc = accuracy_score(y_test, y_pred)
print(test_acc)
```

```python
#making predictive system

input = (1,97,66,15,140,23.2,0.487,22)
array_input = np.asarray(input)
reshaped_input = array_input.reshape(1, -1)
std_data = scaler.transform(reshaped_input)
prediction = model.predict(std_data)
if prediction[0] == 0:
  print('non-diabetic')
else:
  print('diabetic')
```
